```python
# Takes the dataset and uses the rocket column to call the API and append the data to the list
def getBoosterVersion(data):
    for x in data['rocket']:
       if x:
        response = requests.get("https://api.spacexdata.com/v4/rockets/"+str(x)).json()
        BoosterVersion.append(response['name'])
```

```python
# Takes the dataset and uses the cores column to call the API and append the data to the lists
def getCoreData(data):
    for core in data['cores']:
            if core['core'] != None:
                response = requests.get("https://api.spacexdata.com/v4/cores/"+core['core']).json()
                Block.append(response['block'])
                ReusedCount.append(response['reuse_count'])
                Serial.append(response['serial'])
            else:
                Block.append(None)
                ReusedCount.append(None)
                Serial.append(None)
            Outcome.append(str(core['landing_success'])+' '+str(core['landing_type']))
            Flights.append(core['flight'])
            GridFins.append(core['gridfins'])
            Reused.append(core['reused'])
            Legs.append(core['legs'])
            LandingPad.append(core['landpad'])
```

```python
import requests
import pandas as pd

# ---------- resetăm listele (ca să nu se dubleze la rulări repetate) ----------
BoosterVersion = []
PayloadMass = []
Orbit = []
LaunchSite = []
Outcome = []
Flights = []
GridFins = []
Reused = []
Legs = []
LandingPad = []
Block = []
ReusedCount = []
Serial = []
Longitude = []
Latitude = []

# ---------- date statice de rezervă ----------
BACKUP_URL = ("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/"
              "IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
_backup = None
def backup():
    global _backup
    if _backup is None:
        _backup = pd.read_csv(BACKUP_URL)
    return _backup

# ---------- request sigur: întoarce None dacă API-ul nu răspunde corect ----------
_cache = {}
_api_down = False

def api_get(url):
    global _api_down
    if _api_down:
        return None
    if url in _cache:
        return _cache[url]
    try:
        r = requests.get(url, timeout=10)
        if r.status_code == 200 and r.text.strip():
            _cache[url] = r.json()
            return _cache[url]
    except Exception:
        pass
    _api_down = True   # nu mai pierdem timp cu sute de request-uri eșuate
    print("API-ul SpaceX nu răspunde -> folosesc datele statice de rezervă.")
    return None

# ---------- 1. getBoosterVersion ----------
def getBoosterVersion(data):
    for i, x in enumerate(data['rocket']):
        if x:
            resp = api_get("https://api.spacexdata.com/v4/rockets/" + str(x))
            if resp is not None:
                BoosterVersion.append(resp['name'])
            else:
                BoosterVersion.append(backup()['BoosterVersion'].iloc[i])

# ---------- 2. getLaunchSite ----------
def getLaunchSite(data):
    for i, x in enumerate(data['launchpad']):
        if x:
            resp = api_get("https://api.spacexdata.com/v4/launchpads/" + str(x))
            if resp is not None:
                Longitude.append(resp['longitude'])
                Latitude.append(resp['latitude'])
                LaunchSite.append(resp['name'])
            else:
                b = backup()
                Longitude.append(b['Longitude'].iloc[i])
                Latitude.append(b['Latitude'].iloc[i])
                LaunchSite.append(b['LaunchSite'].iloc[i])

# ---------- 3. getPayloadData ----------
def getPayloadData(data):
    for i, load in enumerate(data['payloads']):
        if load:
            resp = api_get("https://api.spacexdata.com/v4/payloads/" + str(load))
            if resp is not None:
                PayloadMass.append(resp['mass_kg'])
                Orbit.append(resp['orbit'])
            else:
                b = backup()
                PayloadMass.append(b['PayloadMass'].iloc[i])
                Orbit.append(b['Orbit'].iloc[i])

# ---------- 4. getCoreData ----------
def getCoreData(data):
    for i, core in enumerate(data['cores']):
        if core['core'] is not None:
            resp = api_get("https://api.spacexdata.com/v4/cores/" + str(core['core']))
            if resp is not None:
                Block.append(resp['block'])
                ReusedCount.append(resp['reuse_count'])
                Serial.append(resp['serial'])
            else:
                b = backup()
                Block.append(b['Block'].iloc[i])
                ReusedCount.append(b['ReusedCount'].iloc[i])
                Serial.append(b['Serial'].iloc[i])
        else:
            Block.append(None)
            ReusedCount.append(None)
            Serial.append(None)
        Outcome.append(str(core['landing_success']) + ' ' + str(core['landing_type']))
        Flights.append(core['flight'])
        GridFins.append(core['gridfins'])
        Reused.append(core['reused'])
        Legs.append(core['legs'])
        LandingPad.append(core['landpad'])
```

```python
_cache = {}
def api_get(url):
    if url not in _cache:
        r = requests.get(url, timeout=10)
        if r.status_code != 200 or not r.text.strip():
            raise RuntimeError(f"API SpaceX indisponibil ({r.status_code}) pentru {url}")
        _cache[url] = r.json()
    return _cache[url]

def getBoosterVersion(data):
    for x in data['rocket']:
        if x:
            BoosterVersion.append(api_get("https://api.spacexdata.com/v4/rockets/" + str(x))['name'])

def getLaunchSite(data):
    for x in data['launchpad']:
        if x:
            r = api_get("https://api.spacexdata.com/v4/launchpads/" + str(x))
            Longitude.append(r['longitude'])
            Latitude.append(r['latitude'])
            LaunchSite.append(r['name'])

def getPayloadData(data):
    for load in data['payloads']:
        if load:
            r = api_get("https://api.spacexdata.com/v4/payloads/" + str(load))
            PayloadMass.append(r['mass_kg'])
            Orbit.append(r['orbit'])

def getCoreData(data):
    for core in data['cores']:
        if core['core'] != None:
            r = api_get("https://api.spacexdata.com/v4/cores/" + core['core'])
            Block.append(r['block'])
            ReusedCount.append(r['reuse_count'])
            Serial.append(r['serial'])
        else:
            Block.append(None)
            ReusedCount.append(None)
            Serial.append(None)
        Outcome.append(str(core['landing_success']) + ' ' + str(core['landing_type']))
        Flights.append(core['flight'])
        GridFins.append(core['gridfins'])
        Reused.append(core['reused'])
        Legs.append(core['legs'])
        LandingPad.append(core['landpad'])
```

```python
try:
    getBoosterVersion(data)
    getLaunchSite(data)
    getPayloadData(data)
    getCoreData(data)
    API_OK = True
except RuntimeError as e:
    print(e)
    API_OK = False
print("API_OK =", API_OK)
```
